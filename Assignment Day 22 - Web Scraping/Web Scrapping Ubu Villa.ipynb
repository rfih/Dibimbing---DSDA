```python
titles = []  # List untuk menyimpan judul

for page in range(1, 4):  # Mengiterasi melalui 4 halaman
    url = f'https://ubuvilla.com/property?page={page}'
  # Ganti URL dengan URL sebenarnya
    response = requests.get(url)
    soup = BeautifulSoup(response.text, 'html.parser') # Parser html documents
    #code diatas ialah untuk multiple page

    #define container terluar
    div_containers = soup.findAll('div', attrs={'class':"property-row col-lg-12 col-md-12 col-sm-12 col-xs-12"})

    for container in div_containers:
        review = container.find('div', attrs={'class': 'post-title'})
        if review:
            titles.append(review.text.strip())

# Membuat dataframe dari list judul
df_villa_title = pd.DataFrame({'title': titles})
print('total properties yang didapat : ', len(df_villa_title)) #validation data
# Menampilkan dataframe
df_villa_title.head()
```

```python
address = []  # List untuk menyimpan judul

for page in range(1, 4):  # Mengiterasi melalui 4 halaman
    url = f'https://ubuvilla.com/property?page={page}'
  # Ganti URL dengan URL sebenarnya
    response = requests.get(url)
    soup = BeautifulSoup(response.text, 'html.parser') # Parser html documents
    #code diatas ialah untuk multiple page

    #define container terluar
    div_containers = soup.findAll('div', attrs={'class':"property-row col-lg-12 col-md-12 col-sm-12 col-xs-12"})

    for container in div_containers:
        review = container.find('ul', attrs={'class': 'property-location'})
        if review:
            address.append(review.text.strip())

# Membuat dataframe dari list judul
df_villa_address = pd.DataFrame({'address': address})
print('total properties yang didapat : ', len(df_villa_address)) #validation data
# Menampilkan dataframe
df_villa_address.head()
```

```python
price = []  # List untuk menyimpan judul

for page in range(1, 4):  # Mengiterasi melalui 4 halaman
    url = f'https://ubuvilla.com/property?page={page}'
  # Ganti URL dengan URL sebenarnya
    response = requests.get(url)
    soup = BeautifulSoup(response.text, 'html.parser') # Parser html documents
    #code diatas ialah untuk multiple page

    #define container terluar
    div_containers = soup.findAll('div', attrs={'class':"property-row col-lg-12 col-md-12 col-sm-12 col-xs-12"})

    for container in div_containers:
        review = container.find('span', attrs={'class': 'property-price'})
        if review:
            price.append(review.text.strip())

# Membuat dataframe dari list judul
df_villa_price = pd.DataFrame({'price': price})
print('total properties yang didapat : ', len(df_villa_price)) #validation data
# Menampilkan dataframe
df_villa_price.head()
```

```python
url1 = "https://ubuvilla.com"
driver1 = webdriver.Firefox()
driver1.get(url1)

soup = BeautifulSoup(driver1.page_source, "html.parser")
```

```python
div_containers1 = soup.findAll('ul', attrs={'class':"post-category-list"})
print(div_containers1)
```

```python
villa_details = []

for ul in div_containers1:
    detail = {}

    # Bedrooms
    bed = ul.find('i', class_='icon-bed2')
    if bed:
        li = bed.find_parent('li')
        text = li.get_text(strip=True)
        match = re.search(r'\d+(\/\d+)?', text)
        detail['Bedrooms'] = match.group(0) if match else None
    else:
        detail['Bedrooms'] = None

    # Bathrooms
    bath = ul.find('i', class_='icon-man-woman')
    if bath:
        li = bath.find_parent('li')
        text = li.get_text(strip=True)
        match = re.search(r'\d+(\/\d+)?', text)
        detail['Bathrooms'] = match.group(0) if match else None
    else:
        detail['Bathrooms'] = None

    # Parking
    detail['Parking'] = 'Yes' if ul.find('i', class_='icon-directions_car') else 'No'

    # Capacity
    cap = ul.find('i', class_='icon-users')
    if cap:
        li = cap.find_parent('li')
        text = li.get_text(strip=True)
        match = re.search(r'\d+(\/\d+)?', text)
        detail['Capacity'] = match.group(0) if match else None
    else:
        detail['Capacity'] = None

    villa_details.append(detail)

df_villa_details = pd.DataFrame(villa_details)
print(f"Total villas extracted: {len(df_villa_details)}")
df_villa_details.head()
```

```python
df_villas = pd.concat([df_villa_title, df_villa_address, df_villa_price, df_villa_details], axis=1)
print('total properties yang didapat : ', len(df_villas))
df_villas.head()
```
