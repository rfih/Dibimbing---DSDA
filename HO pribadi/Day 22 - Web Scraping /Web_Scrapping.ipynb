```python
titles = []  # List untuk menyimpan judul

for page in range(1, 5):  # Mengiterasi melalui 4 halaman
    url = f'https://www.bukitvista.com/property-type/villa/page/{page}'  # Ganti URL dengan URL sebenarnya
    response = requests.get(url)
    soup = BeautifulSoup(response.text, 'html.parser') # Parser html documents
    #code diatas ialah untuk multiple page

    #define container terluar
    div_containers = soup.findAll('div', attrs={'class': 'd-flex align-items-center h-100'})

    for container in div_containers:
        review = container.find('h2', attrs={'class': 'item-title'})
        if review:
            titles.append(review.text.strip())

# Membuat dataframe dari list judul
df_villa_title = pd.DataFrame({'title': titles})
print('total properties yang didapat : ', len(df_villa_title)) #validation data
# Menampilkan dataframe
df_villa_title.head()
```

```python
data_address = []  # List untuk menyimpan judul

for page in range(1, 5):  # Mengiterasi melalui 4 halaman
    url = f'https://www.bukitvista.com/property-type/villa/page/{page}'  # Ganti URL dengan URL sebenarnya
    response = requests.get(url)
    soup = BeautifulSoup(response.text, 'html.parser')
    div_containers = soup.findAll('div', attrs={'class': 'd-flex align-items-center h-100'})

    for container in div_containers:
        address = container.find('address', attrs={"class":'item-address'})
        if address:
            data_address.append(address.text.strip())

# Membuat dataframe dari list judul
df_villa_address = pd.DataFrame({'alamat': data_address})
print('total properties yang didapat : ', len(df_villa_address))
# Menampilkan dataframe
df_villa_address
```

## 2. Terjemahkan proses yang ada pada code `extract_usd` tersebut apa maknanya dan tulis dalam markdown ?

```python
# Konversi angka ke satuan USD
import numpy as np

def extract_usd(value):

    # Cek jika dalam USD

    if "USD" in value or "$" in value:

        usd_match = re.search(r'(\d+[\.,]?\d*)', value)

        if usd_match:

            return float(usd_match.group(1).replace(',', ''))

    # Cek jika dalam Rupiah

    elif "Rp" in value:

        rp_match = re.search(r'Rp([\d\.]+)', value)

        if rp_match:

            rupiah = float(rp_match.group(1).replace('.', ''))

            return np.ceil(rupiah / 16380)  # Konversi ke USD dan bulatkan ke atas

    return np.nan  # Jika tidak sesuai format
```

```python
# Tambahkan kolom baru untuk nilai USD
# df_villa['harga dalam USD'] = df_villa['harga sewa ($ USD)'].apply(extract_usd)
# atau memakai lambda
df_villa['harga dalam USD'] = df_villa['harga sewa ($ USD)'].apply(lambda x: extract_usd(x))
df_villa
```
