### Cara 2 adalah dengan Interpretasi Residual Plot

image.png

```python
import numpy as np
from sklearn.linear_model import LinearRegression

# Contoh data
X = np.random.rand(100, 1)
y = 2 * X.flatten() + np.random.normal(0, 1, 100)

# Fit model
model = LinearRegression().fit(X, y)

# Mendapatkan residuals
y_pred = model.predict(X)
residuals = y - y_pred

# Output fitted values dan residuals
# print("Fitted values:", fitted_values)
# print("Residuals:", residuals)
# fitted_values : data prediksi
# residuals : error
```

```python
plt.scatter(y_pred, residuals)
plt.axhline(y=0, color='r', linestyle='--')
plt.xlabel('Fitted Values')
plt.ylabel('Residuals')
plt.title('Residual Plot')
plt.show()
```

### Tidak Homoskedastisitas atau heteroskedastisitas

image.png

```python
# Contoh data dengan heteroskedastisitas
np.random.seed(0)
X = np.random.rand(100, 1)
y = 3 * X.squeeze() + np.random.randn(100) * (X.squeeze() ** 2)

# Fit model regresi linear
X = sm.add_constant(X)
model = sm.OLS(y, X).fit()
y_pred = model.fittedvalues
residuals = model.resid

# Plot Residual vs Fitted Values
plt.scatter(y_pred, residuals)
plt.axhline(y=0, color='r', linestyle='--')
plt.xlabel('Fitted Values')
plt.ylabel('Residuals')
plt.title('Residuals vs Fitted Values')
plt.show()
```

## Residuals are normally distributed

image.png

```python
import numpy as np
from sklearn.linear_model import LinearRegression

# Contoh data
X = np.random.rand(100, 1)
y = 2 * X.flatten() + np.random.normal(0, 1, 100)

# Fit model
model = LinearRegression().fit(X, y)

# Mendapatkan residuals
fitted_values = model.predict(X)
residuals = y - fitted_values

# Output fitted values dan residuals
# print("Fitted values:", fitted_values)
# print("Residuals:", residuals)
```

```python
sns.distplot(residuals)
```

image.png

```python
# QQ-plot
sm.qqplot(residuals, line='45')
plt.title('QQ-plot Residual (Normalitas)')
plt.show()
```
